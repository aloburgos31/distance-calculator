```python
import math
x_coordinate1=float(input("Enter x coordinate 1: "))
y_coordinate1=float(input("Enter y coordinate 1: "))
x_coordinate2=float(input("Enter x coordinate 2: "))
y_coordinate2=float(input("Enter y coordinate 2: "))
distance=math.sqrt(pow((x_coordinate2-x_coordinate1), 2)+pow((y_coordinate2-y_coordinate1), 2))
print(f"The distance between the points is {distance} units.")

#Why is using a library more practical than writing all calculations from scratch? Explain briefly using your activity as an example.

#Using libraries are so much more practical than writing all calculations from scratch because they take less effort and is easier than manually writing them. Additionally, there are already built in commands that solve different kinds of mathematical equations and problems, such as functions like "sqrt()" and "pow()".
```
